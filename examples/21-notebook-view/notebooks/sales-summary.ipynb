# Sales summary

Quarterly revenue per store, read from `data/sales.csv`. The kernel starts in the project directory, so relative paths work as they do in `odf run`.

```python
import csv
from collections import defaultdict

with open("data/sales.csv", newline="") as f:
    rows = list(csv.DictReader(f))

len(rows)
```

## Revenue per store

```python
totals = defaultdict(int)
for row in rows:
    totals[row["store"]] += int(row["revenue"])

for store, total in sorted(totals.items(), key=lambda item: -item[1]):
    print(f"{store:<6} {total:>8,}")
```

```python
import matplotlib.pyplot as plt

stores = sorted(totals)
fig, ax = plt.subplots(figsize=(6, 3))
ax.bar(stores, [totals[s] for s in stores], color="#4c8bf5")
ax.set_ylabel("Revenue")
ax.set_title("Q3 revenue per store")
plt.show()
```
